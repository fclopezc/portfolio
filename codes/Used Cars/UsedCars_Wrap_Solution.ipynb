## Dealing with Missing values

➡️ 10. For the following variables:

```python
"normalized-losses", "stroke", "bore", "horsepower", "peak-rpm"
```

Replace missing values with the **mean** value of each variable.

🟢 Temporaly, transform them into "float" for the local calculation.

🟠 Use the `axis=0` for get the mean, to get in familiar.

🟠 Remember to use `inplace` for the replacement.

<details><summary>Solution</summary>

```python
col_rep_mean = ["normalized-losses", "stroke", "bore", "horsepower", "peak-rpm"]

for i in col_rep_mean:
    avg_col = df[i].astype('float').mean(axis=0)
    df[i].replace(np.nan, avg_col, inplace= True)
```

➡️ 18. Group the value into 3 levels: Low, Medium, High. For that you need to create four reference point.

2. Group the values into this two.

<details><summary>Solution</summary>

