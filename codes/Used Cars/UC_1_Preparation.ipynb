## Dealing with Missing values

➡️ 10. For the following variables:

```python
"normalized-losses", "stroke", "bore", "horsepower", "peak-rpm"
```

Replace missing values with the **mean** value of each variable.

🟢 Temporaly, transform them into "float" for the local calculation.

🟠 Use the `axis=0` for get the mean, to get in familiar.

🟠 Remember to use `inplace` for the replacement.

<details><summary>Solution</summary>

```python
col_rep_mean = ["normalized-losses", "stroke", "bore", "horsepower", "peak-rpm"]

for i in col_rep_mean:
    df[i] = pd.to_numeric(df[i], errors='coerce')

```


````python
# Answer
col_rep_mean = ["normalized-losses", "stroke", "bore", "horsepower", "peak-rpm"]

for i in col_rep_mean:
    df[i] = pd.to_numeric(df[i], errors='coerce')
    df[i] = df[i].fillna(df[i].mean())
````

```python
df['num-of-doors'].value_counts()
```

📘 15. Transform the variable `horsepower` into integer. Then group the value into 3 levels: Low, Medium, High. For that you need to create four reference point.

1. Create a bins, according to the max and min value for the reference point.
2. Group the values into this two.

<details><summary>Solution</summary>

```python
df["horsepower"] = df["horsepower"].astype("int")

bins = np.linspace(min(df["horsepower"]), max(df["horsepower"]), 4)
group_names = ['Low', 'Medium', 'High']
df['horsepower-binned'] = pd.cut(df['horsepower'],
                                 bins, labels=group_names, include_lowest=True)
df["horsepower-binned"].value_counts()
```

````python
df["horsepower"] = df["horsepower"].astype("int")

bins = np.linspace(min(df["horsepower"]), max(df["horsepower"]), 4)
group_names = ['Low', 'Medium', 'High']
df['horsepower'] = pd.cut(df['horsepower'],
                                 bins, labels=group_names, include_lowest=True)
df["horsepower"].value_counts()
````
